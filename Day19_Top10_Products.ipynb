## 📌 Description

The objective of this task is to identify the top 10 products based on total sales from the Sample Superstore dataset.

The analysis involves grouping sales by product, calculating total sales for each product, sorting the products in descending order, and identifying the top 10 products with the highest sales.

## 🎯 Objective

- Identify the top 10 products based on total sales.
- Aggregate sales at the product level.
- Rank products in descending order of sales.
- Check for ties around the Top 10 cutoff.
- Visualize the Top 10 products using a bar chart.
- Derive meaningful insights from the analysis.

## 1.Import Dataset and Read File

```python
# 1.Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# 2. Read Dataset
df=pd.read_excel('/content/Sample-Superstore2_dataset.xlsx')

# 3.Display the first five rows of the dataset
df.head()
```

## 2.Initial Dataset Inspection

```python
# 4.Display the number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# 5.Display all column names
print("Column Names:")
print(df.columns.tolist())

# 6.Check the data types and non-null values
print(df.info())

# 7.Check for missing values in each column
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

## 3.PRODUCT-WISE SALES ANALYSIS

```python
# 8.Calculate total sales for each product

product_sales = (
    df.groupby("Product Name", as_index=False)["Sales"]
      .sum())
display(product_sales.head())

# 9.Sort products by total sales in descending order

product_sales = product_sales.sort_values(by="Sales",ascending=False)
product_sales.head(10)
```

## 4.Top 10 Products by Sales

The products are ranked based on their total sales, with the highest-selling product appearing first.

```python
# 10.Select the top 10 products

top_10 = product_sales.head(10).copy()

# 11.Add ranking
top_10.insert(0, "Rank", range(1, 11))

display(top_10)

# 12.Round sales values to two decimal places for better readability

top_10["Sales"] = top_10["Sales"].round(2)

display(top_10)
```

## 5.Tie Check

The 10th-ranked product is compared with the 11th-ranked product to determine whether there is a tie at the Top 10 cutoff.

```python
# 13.Display the 10th and 11th ranked products

top_11 = product_sales.iloc[9:11].copy()

top_11.insert(0, "Rank", [10, 11])

top_11
```

```python
# 14.Check whether the 10th and 11th products have equal sales

if top_11.iloc[0]["Sales"] == top_11.iloc[1]["Sales"]:
    print("Tie detected at the Top 10 cutoff.")
else:
    print("No tie detected at the Top 10 cutoff.")
```

##  Combined Sales of Top 10 Products

```python
# 15.Calculate the combined sales of the Top 10 products

total_top10_sales = top_10["Sales"].sum()

print(f"Combined sales of Top 10 products: {total_top10_sales:,.2f}")
```

## Visualization — Top 10 Products by Sales

A horizontal bar chart is used because product names are relatively long and are easier to read horizontally.

```python
# 16.Create a horizontal bar chart for the Top 10 products

plt.figure(figsize=(12, 7))

bars = plt.barh(
    top_10["Product Name"],
    top_10["Sales"]
)

plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.title("Top 10 Products by Sales")

# Display highest-selling product at the top
plt.gca().invert_yaxis()

# Add sales values to the bars
for bar in bars:
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():,.2f}",
        va="center"
    )

plt.tight_layout()
plt.show()
```

##  Key Findings

1. The analysis identified the Top 10 products based on total sales.

4. The products were ranked in descending order based on their total sales.

5. The 10th-ranked product recorded sales of 17,030.31, while the 11th-ranked product recorded sales of 16,829.90.

6. No tie was identified at the Top 10 cutoff.
